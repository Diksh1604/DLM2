# **Ans 4:-**

In TensorFlow 2.0, the Model class serves a very important role — it acts as the core structure for building and training machine learning models.

**Specifically, the purposes of the Model class are:**

Encapsulation of Layers and Computations

A Model is basically a container for layers (tf.keras.layers.Layer objects).

It defines how data flows through layers to produce output.

Training and Evaluation

The Model class provides built-in methods like:

.compile() — for configuring the training process (loss function, optimizer, metrics).

.fit() — for training the model.

.evaluate() — for evaluating the model performance on validation/test data.

.predict() — for making predictions.

Saving and Loading

Models can be saved to disk and reloaded later using .save() and tf.keras.models.load_model().

Subclassing for Custom Models

In addition to simple sequential or functional models, you can subclass tf.keras.Model to create complex architectures with customized behavior (e.g., GANs, custom training loops).

You override the call(self, inputs) method to specify forward pass logic.

Better Organization and Flexibility

It neatly organizes parameters (weights, biases) and ensures easier debugging, transfer learning, and fine-tuning.

```python
#Ans 4

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# Define the model (as before)
model = tf.keras.Sequential([
    tf.keras.layers.InputLayer(input_shape=(784,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compile the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Example data (replace with actual data)
X_train = np.random.rand(1000, 784)
y_train = np.random.randint(2, size=(1000, 1))
X_val = np.random.rand(200, 784)
y_val = np.random.randint(2, size=(200, 1))

# Train the model and capture the training history
history = model.fit(X_train, y_train,
                    epochs=10,
                    batch_size=32,
                    validation_data=(X_val, y_val))

# Step 1: Visualize the training and validation loss
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

# Step 2: Visualize the training and validation accuracy
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Show the plots
plt.tight_layout()
plt.show()
```

```python
#Ans 6

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

# 1. Define the neural network
class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        # Define layers
        self.hidden = nn.Linear(input_size, hidden_size)  # Hidden layer
        self.relu = nn.ReLU()  # ReLU activation function
        self.output = nn.Linear(hidden_size, output_size)  # Output layer

    def forward(self, x):
        # Define forward pass
        x = self.hidden(x)  # Hidden layer
        x = self.relu(x)  # Apply ReLU activation
        x = self.output(x)  # Output layer
        return x

# 2. Instantiate the model
input_size = 784  # Example: input size of 784 for MNIST (28x28 images)
hidden_size = 64  # Number of neurons in the hidden layer
output_size = 1  # For binary classification (output is 1)

model = SimpleNN(input_size, hidden_size, output_size)

# 3. Define loss function and optimizer
criterion = nn.BCEWithLogitsLoss()  # Binary Cross-Entropy Loss (for binary classification)
optimizer = optim.Adam(model.parameters(), lr=0.001)  # Adam optimizer

# 4. Example data (using random data for demonstration)
X_train = np.random.rand(1000, 784).astype(np.float32)  # 1000 samples, 784 features (like MNIST flattened)
y_train = np.random.randint(2, size=(1000, 1)).astype(np.float32)  # Random binary labels (0 or 1)

# Convert data to PyTorch tensors
X_train = torch.tensor(X_train)
y_train = torch.tensor(y_train)

# 5. Train the model
epochs = 10
for epoch in range(epochs):
    model.train()  # Set model to training mode

    # Zero the gradients before running the backward pass
    optimizer.zero_grad()

    # Forward pass: Compute predicted y by passing x to the model
    outputs = model(X_train)

    # Compute the loss
    loss = criterion(outputs, y_train)

    # Backward pass: Compute gradient of the loss w.r.t. model parameters
    loss.backward()

    # Update the weights
    optimizer.step()

    # Print the loss every 100 epochs
    if (epoch+1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

# After training, you can use the model for prediction:
model.eval()  # Set model to evaluation mode
with torch.no_grad():  # Disable gradient calculation during inference
    sample_input = torch.randn(1, input_size)  # Random input for testing
    prediction = model(sample_input)
    print("Prediction:", prediction)
```

```python
#Ans 8

class WeightedMSELoss(nn.Module):
    def __init__(self, weights=None):
        super(WeightedMSELoss, self).__init__()
        self.weights = weights

    def forward(self, outputs, targets):
        # Calculate squared differences
        loss = (outputs - targets) ** 2

        # Apply weights (if provided)
        if self.weights is not None:
            loss = loss * self.weights

        return torch.mean(loss)

# Example usage
model_output = torch.tensor([0.5, 0.2, 0.7])  # Predicted output
target = torch.tensor([1.0, 0.0, 1.0])  # True labels
weights = torch.tensor([0.5, 2.0, 1.0])  # Weights for each sample

# Instantiate the custom loss function with weights
loss_fn = WeightedMSELoss(weights=weights)

# Compute the loss
loss = loss_fn(model_output, target)
print("Weighted MSE Loss:", loss.item())
```
